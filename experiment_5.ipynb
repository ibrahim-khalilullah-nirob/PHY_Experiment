# Experiment No: 05
## Determination of the angle of a prism (by rotation of the telescope)

### **Calculations and Formulas:**
* **Total Reading ($P$ or $N$)** = $\text{Main Scale Reading } (S) + (\text{Vernier Scale Division } (VD) \times \text{Vernier Constant } (VC))$
* **AB side angle ($M$)** = $360^\circ - P$
* **AC side angle ($N$)** = $\text{Total reading for face AC}$
* **Total angle between the two rays ($\theta$)** = $M + N$
* **Angle of the prism ($A$)** = $\frac{\theta}{2}$
* **Percentage Error** = $\frac{|\text{Standard Value} - \text{Experimental Value}|}{\text{Standard Value}} \times 100\%$

```python
import numpy as np
import pandas as pd

# Experimental Data Table based on your sheet
data = {
    'Obs': [1, 2, 3],
    'AB_Main (s)': [299, 298, 297],
    'AB_VD': [12, 25, 20],
    'AB_Total (P)': [299.2, 298.41, 297.33],
    'AB_Angle (M)': [60.8, 61.59, 62.67],
    'AC_Main (s)': [60, 57, 58],
    'AC_VD': [25, 19, 20],
    'AC_Total (N)': [60.41, 57.31, 58.33],
    'Total Angle theta': [121.21, 118.9, 121],
    'Mean theta': [120.37, 120.37, 120.37],
    'Prism Angle A (deg)': [60.185, 60.185, 60.185]
}

df = pd.DataFrame(data)

print("=" * 70)
print("EXPERIMENT NO: 05 - DATA TABLE")
print("=" * 70)
print(df.to_string(index=False))
print("=" * 70)

# Error Calculation
standard_value = 60.0  # Standard angle of prism (e.g., 60 degree)
experimental_value = 60.185

percentage_error = (abs(standard_value - experimental_value) / standard_value) * 100

print("\nFINAL RESULTS:")
print(f" * Calculated Prism Angle (A) : {experimental_value}°")
print(f" * Percentage Error          : {percentage_error:.3f}%")
print("=" * 70)
```
