```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tabulate import tabulate

# 1. Observation Table
load_M = [50, 100, 150, 200]
extension = [2.7, 10.0, 18.5, 26.6]

data = {
    "Observation No.": [1, 2, 3, 4],
    "Load M (g)": load_M,
    "Extension (cm)": extension
}

df = pd.DataFrame(data)
print("--- Observation Table ---")
print(tabulate(df, headers="keys", tablefmt="grid", showindex=False))

# 2. Plotting Graph
load_M_arr = np.array(load_M)
extension_arr = np.array(extension)

delta_mass = 15       # g
delta_extension = 2.5 # cm
graph_slope = delta_mass / delta_extension # g/cm

intercept = np.mean(load_M_arr - graph_slope * extension_arr)

extension_fit = np.linspace(min(extension_arr), max(extension_arr), 100)
load_fit = graph_slope * extension_fit + intercept

plt.figure(figsize=(9, 5))
plt.scatter(extension_arr, load_M_arr, color="blue", s=60, label="Experimental Data")
plt.plot(extension_fit, load_fit, color="red", linewidth=2, label="Best-fit Straight Line")

plt.xlabel("Extension, l (cm)")
plt.ylabel("Load, M (g)")
plt.title("Dependency of Load on Extension")
plt.grid(True, alpha=0.4)
plt.legend()

# Write your ID
plt.text(0.03, 0.95, "ID : 2025200000246", transform=plt.gca().transAxes, verticalalignment="top")
plt.show()

# 3. Spring Constant Calculation (K)
# Formula: Slope = ΔMass / ΔExtension
# Formula: K = Slope * g
g = 980  # cm/s^2
slope = delta_mass / delta_extension
K = slope * g

print("--- Spring Constant Calculation ---")
print(f"Calculated Slope = {slope:.2f} g/cm")
print(f"Spring Constant (K) = {K:.2f} dynes/cm")
```

```python
# 4. Geometry Data (Vernier Calipers & Screw Gauge)
# Formula: D = MS_ext + (VS_ext * VC)
# Formula: d = MS_int + (VS_int * VC)
# Formula: R = (D + d) / 4
VC = 0.005
MS_ext, VS_ext = 2.3, 0.85
MS_int, VS_int = 2.2, 0.4

D = MS_ext + (VS_ext * VC)
d = MS_int + (VS_int * VC)
R = (D + d) / 4.0

# Formula: wire_diameter = LS + (CS * LC)
# Formula: r = wire_diameter / 2
LC = 0.01
LS, CS = 0.09, 0.9

wire_diameter = LS + (CS * LC)
r = wire_diameter / 2.0

# 5. Rigidity Modulus (η)
# Formula: η = (4 * N * R^3 * K) / r^4
N = 108  # Number of turns
eta = (4 * N * (R**3) * K) / (r**4)

print("--- Geometry Results ---")
print(f"External Diameter (D) = {D:.5f} cm")
print(f"Internal Diameter (d) = {d:.5f} cm")
print(f"Mean Radius (R) = {R:.5f} cm")
print(f"Wire Diameter = {wire_diameter:.3f} cm")
print(f"Wire Radius (r) = {r:.4f} cm")

print("\n--- Rigidity Modulus Result ---")
print(f"Rigidity Modulus (η) = {eta:.2e} dynes/cm²")
```
