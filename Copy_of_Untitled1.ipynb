# Experiment No: 05
## Determination of the Angle of a Prism (by Rotation of the Telescope)

### Theory
Total Reading for Face AB:
$$P = S + (VD \times VC)$$

Reflected Angle for Face AB:
$$M = 360^\circ - P$$

Total Reading for Face AC:
$$N = S + (VD \times VC)$$

Total Angle Between Two Reflected Rays:
$$\theta = M + N$$

Angle of the Prism:
$$A = \frac{\theta}{2}$$

Percentage Error:
$$\text{Percentage Error} = \frac{|A_{\text{theoretical}} - A_{\text{experimental}}|}{A_{\text{theoretical}}} \times 100\%$$

Where symbols have their usual meanings ($S = \text{Main Scale Reading}$, $VD = \text{Vernier Division}$, $VC = \text{Vernier Constant}$).

```python
import numpy as np
import pandas as pd
from tabulate import tabulate

# Vernier Constant (1 arcminute = 1/60 degrees)
VC = 1 / 60  # in degrees

# ========== OBSERVATION DATA ==========
# Reading for Face AB
S_AB = [299, 298, 297]
VD_AB = [12, 25, 20]

# Reading for Face AC
S_AC = [60, 57, 58]
VD_AC = [25, 19, 20]

# ========== CALCULATIONS ==========
# Formula: P = S + (VD * VC)
# Formula: M = 360 - P
P = [s + (vd * VC) for s, vd in zip(S_AB, VD_AB)]
M = [360.0 - p for p in P]

# Formula: N = S + (VD * VC)
# Formula: theta = M + N
N = [s + (vd * VC) for s, vd in zip(S_AC, VD_AC)]
theta = [m + n for m, n in zip(M, N)]

# ========== OBSERVATION TABLE ==========
data = {
    "Obs No.": [1, 2, 3],
    "Main Scale AB (s)": S_AB,
    "VD AB": VD_AB,
    "Total P (deg)": [round(p, 2) for p in P],
    "AB Angle M (deg)": [round(m, 2) for m in M],
    "Main Scale AC (s)": S_AC,
    "VD AC": VD_AC,
    "AC Angle N (deg)": [round(n, 2) for n in N],
    "Total Angle θ (deg)": [round(th, 2) for th in theta]
}

df = pd.DataFrame(data)
print("--- Observation Table ---")
print(tabulate(df, headers="keys", tablefmt="grid", showindex=False))

# ========== RESULTS & ERROR ANALYSIS ==========
# Formula: Mean θ = sum(theta) / len(theta)
# Formula: A = Mean θ / 2
# Formula: Percentage Error = (|A_theoretical - A_exp| / A_theoretical) * 100

mean_theta = np.mean(theta)
A_exp = mean_theta / 2.0
A_theoretical = 60.0  # Standard angle for an equilateral prism

percentage_error = (abs(A_theoretical - A_exp) / A_theoretical) * 100

print("\n--- Final Calculation Results ---")
print(f"Mean θ = {mean_theta:.2f}°")
print(f"Angle of the Prism (A) = {A_exp:.3f}°")

print("\n--- Error Analysis ---")
print(f"Percentage Error = {percentage_error:.3f}%")
```
