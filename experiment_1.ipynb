## Theory

A meter bridge is based on the principle of the Wheatstone bridge. At the balance point, no current flows through the galvanometer. Therefore, the ratio of the resistances is equal to the ratio of the corresponding lengths of the bridge wire.

At balance condition,

$$
\frac{X}{R}=\frac{l}{100-l}
$$

Hence,

$$
X=\frac{Rl}{100-l}
$$

The specific resistance (resistivity) of the wire is

$$
\rho=\frac{X\pi r^2}{L}
$$

Where,

- **X** = Unknown resistance (Ω)
- **R** = Known resistance (Ω)
- **l** = Balance length (cm)
- **r** = Radius of the wire (cm)
- **L** = Length of the wire (cm)
- **ρ** = Specific resistance (Ω·cm)

## Calculation

Using the formula,

$$
X=\frac{Rl}{100-l}
$$

and,

$$
\rho=\frac{X\pi r^2}{L}
$$

Substitute the observed values to calculate the unknown resistance and the specific resistance of the given wire.

```python
from math import pi

# ===== Enter Experimental Values =====
R = 4.0      # Known resistance (Ω)
l = 0.0      # Balance length (cm)
r = 0.0      # Radius of the wire (cm)
L = 1.0      # Length of the wire (cm) (Replace with actual value)

# ===== Calculation =====
X = (R * l) / (100 - l)

rho = (X * pi * r**2) / L

# ===== Output =====
print(f"Unknown Resistance (X) = {X:.4f} Ω")
print(f"Specific Resistance (ρ) = {rho:.6f} Ω·cm")
```
