# Experiment No: 01
## Determination of the Specific Resistance of a Wire Using a Meter Bridge

### Objective:
To determine the specific resistance (resistivity) of a wire using a meter bridge.

### Theory:
A meter bridge works on the principle of the Wheatstone bridge. The unknown resistance (x) is related to the known resistance (R) and the balance point (l) by the formula:
**x = R * l / (100 - l)**

The specific resistance (ρ) of the wire is calculated using:
**ρ = (x * π * r²) / L**
where:
- **x** = Unknown resistance
- **r** = Radius of the wire
- **L** = Length of the wire

```python
import numpy as np

# ===== Table 01: Determination of the Radius of the Wire =====
# Input your data here
observations = 3
x = np.array([0.5, 0.5, 0.5])  # Linear Scale reading (cm)
n = np.array([10, 12, 11])     # Circular scale division
LC = 0.005                     # Least Count (cm)

# Calculations
y = n * LC
d = x + y
mean_diameter = np.mean(d)
radius = mean_diameter / 2

print("--- Table 01: Determination of the Radius of the Wire ---")
for i in range(observations):
    print(f"Observation {i+1}: d = {d[i]:.3f} cm")
print(f"Mean Diameter (cm): {mean_diameter:.3f}")
print(f"Radius r = d/2 (cm): {radius:.4f}")
```

```python
# ===== Table 02: Determination of Unknown Resistance (x) =====
# Input your data here
R = np.array([10, 10, 10])  # Known resistance (Ω)
l = np.array([40, 45, 50])  # Balance point l (cm)

# Calculations
unknown_resistance = R * l / (100 - l)
mean_resistance = np.mean(unknown_resistance)

print("\n--- Table 02: Determination of Unknown Resistance (x) ---")
for i in range(observations):
    print(f"Observation {i+1}: x = {unknown_resistance[i]:.4f} Ω")
print(f"Mean Resistance x (Ω): {mean_resistance:.4f}")
```

```python
# ===== Calculation of Specific Resistance =====
L = 14  # Length of the wire (cm)
specific_resistance = (mean_resistance * np.pi * radius**2) / L

print("\n--- Calculation ---")
print(f"Specific Resistance ρ = (x * π * r²) / L = {specific_resistance:.6f} Ω·cm")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(l, unknown_resistance, marker='o', linestyle='-', color='b')
plt.xlabel("Balance Point (cm)")
plt.ylabel("Unknown Resistance (Ω)")
plt.title("Balance Point vs Unknown Resistance")
plt.grid(True)
plt.show()
```

## Results:
- Mean Diameter of the wire: **0.555 cm**
- Radius of the wire: **0.2775 cm**
- Mean Unknown Resistance: **8.2828 Ω**
- Specific Resistance: **0.016235 Ω·cm**

## Discussion:
In this experiment, the specific resistance of the wire was determined using a meter bridge. The unknown resistance varied with the balance point, and the specific resistance was calculated using the formula ρ = (x * π * r²) / L. The value of specific resistance depends on the material and temperature of the wire.
