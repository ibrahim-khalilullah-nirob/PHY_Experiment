## Determination of the Spring Constant and Rigidity Modulus of the Material of a Spring

### Introduction

The aim of this experiment is to determine the spring constant and the rigidity modulus of the material of a given spring.

### Theory

According to Hooke's law,

$$F = Kl$$

For a load of mass $M$,

$$Mg = Kl$$

Therefore,

$$M = \frac{K}{g}l$$

If load $M$ is plotted along the y-axis and extension $l$ along the x-axis, then

$$\text{Slope}=\frac{\Delta M}{\Delta l}=\frac{K}{g}$$

Therefore,

$$K=\text{Slope}\times g$$

The rigidity modulus of the material of the spring is,

$$\eta=\frac{4NR^3K}{r^4}$$

Where,

- $N$ = Number of turns of the spring
- $R$ = Mean radius of the spring
- $r$ = Radius of the spring wire
- $K$ = Spring constant

```python
import numpy as np
import matplotlib.pyplot as plt

load_M = np.array([50, 100, 150, 200])
extension = np.array([2.7, 10.0, 18.5, 26.6])

# Slope obtained manually from the graph
graph_slope = 15 / 2.5  # 6 g/cm

# Intercept for drawing the straight line
intercept = np.mean(load_M - graph_slope * extension)

# Values for the straight line
extension_fit = np.linspace(
    min(extension),
    max(extension),
    100
)

load_fit = graph_slope * extension_fit + intercept

# Draw graph
plt.figure(figsize=(10, 6))

# Experimental data points
plt.scatter(
    extension,
    load_M,
    color="blue",
    s=60,
    label="Experimental Data"
)

# Straight line
plt.plot(
    extension_fit,
    load_fit,
    color="red",
    linewidth=2,
    label="Best-fit Straight Line"
)

plt.xlabel("Extension, l (cm)")
plt.ylabel("Load, M (g)")
plt.title("Dependency of Load on Extension")

plt.grid(True, alpha=0.4)
plt.legend()

# Write your ID, batch and section
plt.text(
    0.03,
    0.95,
    "ID : 2025200000246",
    transform=plt.gca().transAxes,
    verticalalignment="top"
)

plt.show()
```

### Calculation of Spring Constant

The load is taken along the y-axis and the extension is taken along the x-axis.

The slope and intercept of the best-fit straight line are calculated using `np.polyfit()`.

$$K=\text{Slope}\times g$$


```python
# Values obtained from the graph
delta_mass = 15       # g
delta_extension = 2.5 # cm
g = 980               # cm/s²

# Calculate slope
slope = delta_mass / delta_extension

# Calculate spring constant
K = slope * g

print("Slope for Spring Constant:")
print("Slope = ΔMass / ΔExtension")
print(f"Slope = {delta_mass} / {delta_extension}")
print(f"Slope = {slope:.2f} g/cm")

print("\nSpring Constant:")
print("K = Slope × g")
print(f"K = {slope:.2f} × {g}")
print(f"K = {K:.2f} dynes/cm")
```

```python
VC = 0.005  # Vernier constant in cm

# External diameter
MS_ext = 2.3
VS_ext = 0.85

D = MS_ext + (VS_ext * VC)

# Internal diameter
MS_int = 2.2
VS_int = 0.4

d = MS_int + (VS_int * VC)

# Mean radius
R = (D + d) / 4

print(f"External Diameter, D = {D:.5f} cm")
print(f"Internal Diameter, d = {d:.5f} cm")

print("\nR = (D + d) / 4")
print(f"R = ({D:.5f} + {d:.5f}) / 4")
print(f"Mean Radius, R = {R:.5f} cm")
```

```python
LC = 0.01  # Least count in cm

LS = 0.09
CS = 0.9

wire_diameter = LS + (CS * LC)
r = wire_diameter / 2

print(f"Wire Diameter = {wire_diameter:.3f} cm")

print("\nr = Wire Diameter / 2")
print(f"r = {wire_diameter:.3f} / 2")
print(f"Wire Radius, r = {r:.4f} cm")
```

```python
N = 108  # Number of turns

eta = (4 * N * R**3 * K) / r**4

print("η = 4NR³K / r⁴")
print(
    f"η = [4 × {N} × ({R:.5f})³ × {K:.2f}] "
    f"/ ({r:.4f})⁴"
)

print(f"Rigidity Modulus, η = {eta:.2e} dynes/cm²")
```
