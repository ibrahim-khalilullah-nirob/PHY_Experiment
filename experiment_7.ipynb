# Determination of Spring Constant and Rigidity Modulus of the Material of a Given Spring

```python
import numpy as np
import pandas as pd

# Creating DataFrame for Experimental Data
data = {
    'Observation No': [1, 2, 3, 4],
    'Loads M (gm)': [50, 100, 150, 200],
    'Extension cm': [6.7, 10.0, 18.5, 26.6],
    'No of vibrations n': [10, 10, 10, 10],
    'Total time s': [7, 9, 11, 13],
    'Period T = t/n (s)': [0.7, 0.9, 1.1, 1.3],
    'T^2 (s^2)': [0.49, 0.81, 1.21, 1.69]
}

df = pd.DataFrame(data)

print("=" * 65)
print("EXPERIMENT DATA TABLE")
print("=" * 65)
print(df.to_string(index=False))
print("=" * 65)
```

### **Calculations:**

* **External diameter D** = M.S + (V.S $\times$ V.C) = $2.3 + (85 \times 0.005) = 2.30425\text{ cm}$

* **Internal diameter d** = M.S + (V.S $\times$ V.C) = $2.2 + (9 \times 0.005) = 2.202\text{ cm}$

* **Radius of the spring $R$** = $\frac{D + d}{4} = \frac{2.30425 + 2.202}{4} = 1.12656\text{ cm}$

* **Diameter of the wire, $d'$** = L.S + (C.S $\times$ L.C) = $0.5 + (49 \times 0.01) = 0.99\text{ mm} = 0.099\text{ cm}$

* **Radius of the wire, $r$** = $\frac{d'}{2} = \frac{0.099}{2} = 0.0495\text{ cm}$

* **No of turns of the spring, $N$** = $108$

$$\text{Rigidity, } \eta = \frac{4NR^3K}{r^4}\text{ dynes/cm}^2$$

```python
import numpy as np

# Given parameters from your calculation sheet
N = 108
R = 1.12656
r = 0.0495
K = 5880

# Formula implementation
eta = (4 * N * (R**3) * K) / (r**4)

# Output Box Display
print("=" * 60)
print("FINAL RIGIDITY MODULUS CALCULATION")
print("=" * 60)
print(f"Rigidity Modulus (eta) = {eta:.2e} dynes/cm^2")
print("=" * 60)
```
